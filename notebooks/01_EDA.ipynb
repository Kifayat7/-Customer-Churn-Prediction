Understand the Telco customer base and what drives churn.

```python
import sys; sys.path.insert(0, "..")
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from src.data_preprocessing import *
sns.set_theme(style="whitegrid")
```

```python
df = load_data()
print(df.shape)
df.head()
```

```python
df.info()
df.isna().sum().sum(), df.duplicated().sum()
```

`TotalCharges` is stored as text (blank for brand-new customers) - fixed in notebook 02.

```python
df = clean_data(df)
df.describe().T
```

```python
print(df['Churn'].value_counts(normalize=True).round(3))
sns.countplot(data=df, x='Churn', hue='Churn', legend=False);
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, c in zip(axes, ['Contract', 'InternetService', 'PaymentMethod']):
    (df.groupby(c)['Churn'].apply(lambda s: (s == 'Yes').mean() * 100)).plot.bar(ax=ax, title=f'Churn % by {c}')
plt.tight_layout()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, c in zip(axes, ['tenure', 'MonthlyCharges', 'TotalCharges']):
    sns.histplot(data=df, x=c, hue='Churn', kde=True, ax=ax)
plt.tight_layout()
```

```python
num = df[['SeniorCitizen','tenure','MonthlyCharges','TotalCharges']].assign(Churn=(df.Churn=='Yes').astype(int))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm');
```

**Findings:** month-to-month contracts, fiber-optic internet, electronic-check payment and short tenure show the highest churn.
