```python
import sys; sys.path.insert(0, "..")
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from src.data_preprocessing import *
sns.set_theme(style="whitegrid")
```

```python
from sklearn.pipeline import Pipeline
from src.train import get_models
from src.evaluation import compute_metrics
X, y = prepare_xy(load_data())
X_train, X_test, y_train, y_test = split_data(X, y)
X_train.shape, X_test.shape
```

```python
rows = {}
for name, clf in get_models().items():
    pipe = Pipeline([('prep', build_preprocessor()), ('clf', clf)]).fit(X_train, y_train)
    prob = pipe.predict_proba(X_test)[:, 1]
    rows[name] = compute_metrics(y_test, (prob >= .5).astype(int), prob)
pd.DataFrame(rows).T.round(4)
```

Run `python -m src.train` to train, pick the best model by ROC-AUC and save it to `models/churn_model.pkl`.
